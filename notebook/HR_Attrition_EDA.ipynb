# HR Attrition - Data Cleaning & Initial Exploration

Dataset: IBM HR Analytics Employee Attrition (Kaggle), 1470 rows, 35 columns.

This notebook covers loading the raw CSV, checking data quality, dropping
columns with no analytical value, and a first exploratory pass before
moving the cleaned data into SQL Server for the deeper analysis.

## Load the data

```python
import pandas as pd

df = pd.read_csv('WA_Fn-UseC_-HR-Employee-Attrition.csv')
df.head()
```

## Basic shape and data types

```python
print("Shape (rows, columns):", df.shape)
print("\nColumn names and data types:")
print(df.dtypes)
```

```python
df.describe()
```

## Check categorical columns

Quick look at how many departments and job roles are in the data.

```python
print(df['Department'].value_counts())
print()
print(df['JobRole'].value_counts())
```

## Drop columns with no analytical value

`EmployeeCount`, `Over18`, and `StandardHours` never change for
any row -- checked with `.nunique()`. Not useful for anything, so dropping them.

```python
print("Unique values per column (checking for constants):")
for col in ['EmployeeCount', 'Over18', 'StandardHours']:
    print(col, '->', df[col].unique())
```

```python
df = df.drop(columns=['EmployeeCount', 'Over18', 'StandardHours'])
print("New shape:", df.shape)
```

## Check for duplicate rows

```python
print("Duplicate rows:", df.duplicated().sum())
```

## Check for missing values

(Turned out to be zero -- surprised me a bit, most datasets have at least a few.)

```python
df.isnull().sum().sum()
```

## Overall attrition rate

```python
attrition_counts = df['Attrition'].value_counts()
attrition_rate = df['Attrition'].value_counts(normalize=True) * 100

print("Counts:\n", attrition_counts)
print("\nPercentage:\n", attrition_rate.round(2))
```

## Attrition rate by department

```python
dept_attrition = df.groupby('Department')['Attrition'].value_counts(normalize=True) * 100
print(dept_attrition.round(2))
```

## Comparing employees who left vs stayed

Average income, job satisfaction, work-life balance, and tenure,
split by attrition status.

```python
comparison = df.groupby('Attrition')[['MonthlyIncome', 'JobSatisfaction', 'WorkLifeBalance', 'YearsAtCompany']].mean()
print(comparison.round(2))
```

## OverTime vs attrition

```python
overtime_attrition = pd.crosstab(df['OverTime'], df['Attrition'], normalize='index') * 100
print(overtime_attrition.round(2))
```

## Export cleaned data

This CSV is what gets loaded into SQL Server for the rest of the
analysis (see `schema.sql` and `hr_analysis_queries.sql`).

```python
df.to_csv('HR_Attrition_Cleaned.csv', index=False)
print("Saved successfully!")
```
